Dado que la investigación se centra en los departamentos y comunas que integran el Área Metropolitana de Buenos Aires (AMBA), se filtró el dataset original conservando únicamente las observaciones correspondientes a dicha región.

```python
print(df_censo["CODDPTO"].dtype)
print(df_censo["CODDPTO"].head())
```

Antes de realizar cualquier filtrado, es necesario inspeccionar la estructura de la columna `CODDPTO` (código de departamento).

```python
df_censo_amba = df_censo[df_censo["CODDPTO"].isin(amba_departamentos)]
df_censo_amba.head()
##Utilizamos el método .isin() de Pandas. Esta función es más eficiente para escribir múltiples condiciones
# Al pasar nuestra lista `amba_departamentos` como argumento, Pandas  selecciona solo las filas que coinciden con nuestros códigos de interés.
# Este nuevo DataFrame (`df_censo_amba`) será la base para todos nuestros cálculos estadísticos posteriores.
```

Finalmente aplico los filtros para obtener unicamente informacion de los departamentos del amba

```python
df_censo_amba.tail()
```

```python
#Creo una copia de "df_censo_amba" para que los cambios no afecten al dataset principal
df_censo_amba = df_censo[df_censo["CODDPTO"].isin(amba_departamentos)].copy()
```

```python
#porcentaje de personas con estudios universitarios completos sobre el total de población de 18 años o más.
df_censo_amba["pct_universitarios"] = df_censo_amba["p18a_universitaria"] / df_censo_amba["p_18a"] * 100

#porcentaje de personas cuyo máximo nivel educativo alcanzado corresponde a la educación secundaria completa
df_censo_amba["pct_solosecundaria"] = df_censo_amba["p18a_solosecundaria"] / df_censo_amba["p_18a"] * 100
#porcentaje de personas cuyo maximo nivel educativo alcanzado fue primario y sin instruccionn formal.
df_censo_amba["pct_baja_educacion"]=(df_censo_amba["p18a_sin_instruccion"] + df_censo_amba["p18a_soloprimaria"]) /df_censo_amba["p_18a"] * 100
```

Las variables originales contienen cantidades absolutas de personas según nivel educativo. Para permitir comparaciones entre departamentos con poblaciones de distinto tamaño, se construyeron indicadores porcentuales utilizando como denominador la población de 18 años o más.

Se construyó un indicador de baja educación sumando las personas sin instrucción formal y aquellas cuyo máximo nivel alcanzado corresponde a la educación primaria.
